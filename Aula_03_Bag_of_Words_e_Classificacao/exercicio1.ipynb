```python
# Exercício 1 - Construção de Bag of Words para classificação de SMS
# Objetivo: Criar a representação vetorial (BoW) de mensagens SMS (spam e legítimas)

# Importa o CountVectorizer para transformar texto em vetores de contagem de palavras
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# Corpus de treino: 10 mensagens SMS (5 spam + 5 legítimas/ham)
sms_treino = [
    "Parabéns! Você ganhou um prêmio exclusivo, clique no link para resgatar agora.",
    "Ganhe dinheiro rápido trabalhando de casa, cadastre-se já!",
    "Seu número foi sorteado! Responda SIM para receber o prêmio.",
    "Oferta imperdível! Compre agora e ganhe 50% de desconto.",
    "Clique neste link e receba seu bônus imediatamente.",
    "Oi, tudo bem? Como você está?",
    "Vamos almoçar juntos amanhã?",
    "Cheguei em casa, depois te ligo.",
    "Não esqueça da reunião às 14h.",
    "Pode me enviar o endereço por mensagem?"
]

# Inicializa o vetorizador que faz a tokenização e contagem automaticamente
vectorizer = CountVectorizer()

# Ajusta o vocabulário e transforma as mensagens em vetores de contagem
X = vectorizer.fit_transform(sms_treino)

# Exibe o vocabulário aprendido (todas as palavras únicas encontradas)
print("Vocabulário:")
print(vectorizer.get_feature_names_out())

# Exibe a matriz Bag of Words (cada linha = mensagem, cada coluna = palavra)
print("\nVetores:")
print(X.toarray())

```
