### Inicializando o Modelo

### python -m spacy download pt_core_news_sm (rodar no terminal - instala o modelo) 

```python
import spacy

nlp = spacy.load("pt_core_news_sm")

texto = "O professor Tiago está ensianndo PLN na Fatec Jahu."
doc = nlp(texto)
```

### Pipeline de analise do texto

```python
print(nlp.pipe_names)
```

### Obtendo os tokens do texto

```python
for token in doc:
    print(token.text)
```

### Realizando a Lematização

```python
#Imprimindo os lemas dos tokens
# ""O professor Tiago explicou PLN usando spaCy na Fatec Jahu."
for token in doc:
    print(token.text, "->", token.lemma_)
```

### Analisando a classe gramatical dos tokens

```python
# Imprimindo suas classes grmaticais
# "O professor Tiago explicou PLN usando spaCy na Fatec Jahu."

for token in doc:
    print(f"{token.text:10} -> {token.pos_}")
```

### Identificando as Stopwords

```python
# Imprimindo os tokens e se são stop words
for token in doc:
    print(f"{token.text:10} -> {token.is_stop}")
```

### Dependência Sintática

```python
# Imprimindo as dependências sintáticas dos tokens
for token in doc:
    print(f"{token.text:10} -> {token.dep_:10} -> {token.head.text}")
    
```

### NER - Obtenção das Entidades num texto

```python
for ent in doc.ents:
    print(f"{ent.text:10} -> {ent.label_}")
```

## Vetores e Similaridade (word embedings)

```python
nlp = spacy.load("pt_core_news_lg")

doc1 = nlp("inteligência artificial")
doc2 = nlp("aprendizado de máquina")

similarity = doc1.similarity(doc2)
print(f"Similaridade entre '{doc1.text}' e '{doc2.text}': {(similarity*100):.2f}%")
```

## Análise de sentenças

```python
# Imprimindo as sentenças do texto
doc = nlp("Eu gosto de IA. Ele transforma o mundo")

for sent in doc.sents:
    print(sent.text)

```
