```python
# Exercício 3 - Bag of Words com remoção de Stop Words em português
# Objetivo: Repetir o exercício 2, mas removendo stop words para obter palavras mais relevantes

# Importações necessárias
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# Mesmo texto do exercício 2 para comparação
texto = """
O Brasil é um país muito grande e o Brasil possui muitas cidades.
O país tem uma economia forte e muitas oportunidades de trabalho.
"""

# Lista simples de stop words em português
# (palavras muito comuns que não carregam significado relevante)
stop_words_pt = [
    "o","a","os","as","de","do","da","dos","das",
    "em","no","na","nos","nas","um","uma","uns","umas",
    "e","é","que","para","com","por"
]

# Inicializa o vetorizador passando a lista de stop words para remoção
vectorizer = CountVectorizer(stop_words=stop_words_pt)

# Transforma o texto em vetores, ignorando as stop words
X = vectorizer.fit_transform([texto])

# Obtém o vocabulário resultante (sem stop words)
palavras = vectorizer.get_feature_names_out()

# Soma as contagens por palavra
frequencia = X.toarray().sum(axis=0)

# Cria DataFrame para visualização
df = pd.DataFrame({
    "palavra": palavras,
    "frequencia": frequencia
})

# Exibe todas as palavras ordenadas por frequência (sem stop words)
print(df.sort_values(by="frequencia", ascending=False))
```
