```python
# Exercício 2 - Análise de frequência de palavras com Bag of Words
# Objetivo: Criar um BoW de um texto e rankear as palavras mais frequentes

# Importa o CountVectorizer para vetorização de texto
from sklearn.feature_extraction.text import CountVectorizer
# Importa pandas para manipulação de dados tabulares
import pandas as pd


# Texto de exemplo sobre o Brasil
texto = """
O Brasil é um país muito grande e o Brasil possui muitas cidades.
O país tem uma economia forte e muitas oportunidades de trabalho.
"""

# Inicializa o vetorizador (sem remoção de stop words neste exercício)
vectorizer = CountVectorizer()

# Transforma o texto em uma matriz de contagem de termos
X = vectorizer.fit_transform([texto])

# Obtém a lista de palavras do vocabulário
palavras = vectorizer.get_feature_names_out()

# Soma as contagens de cada palavra (eixo 0 = por coluna)
frequencia = X.toarray().sum(axis=0)

# Cria um DataFrame para visualização organizada
df = pd.DataFrame({
    "palavra": palavras,
    "frequencia": frequencia
})

# Exibe as 10 palavras mais frequentes em ordem decrescente
print(df.sort_values(by="frequencia", ascending=False).head(10))
```
