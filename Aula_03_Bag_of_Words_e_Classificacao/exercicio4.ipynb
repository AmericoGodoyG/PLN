```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
import pandas as pd

sms = [
    "Ganhe dinheiro agora clicando neste link",
    "Parabéns você ganhou um prêmio exclusivo",
    "Promoção imperdível clique no link agora",
    "Você foi selecionado para ganhar dinheiro",
    "Resgate seu prêmio agora mesmo",

    "Oi tudo bem com você",
    "Vamos almoçar amanhã",
    "Te ligo mais tarde",
    "Não esqueça da reunião hoje",
    "Cheguei em casa já"
]

labels = [
    "spam","spam","spam","spam","spam",
    "ham","ham","ham","ham","ham"
]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(sms)

# Vocabulário
palavras = vectorizer.get_feature_names_out()

print("VOCABULÁRIO:")
print(palavras)
print()


print("MATRIZ BAG OF WORDS:")
print(X.toarray())
print()

# -----------------------------
# 3. Contagem total de palavras
# -----------------------------

frequencia = X.toarray().sum(axis=0)

df = pd.DataFrame({
    "palavra": palavras,
    "frequencia": frequencia
})

print("PALAVRAS MAIS FREQUENTES:")
print(df.sort_values(by="frequencia", ascending=False))
print()

# -----------------------------
# 4. Treinar modelo Naive Bayes
# -----------------------------

modelo = MultinomialNB()

modelo.fit(X, labels)

print("MODELO TREINADO")
print()

# -----------------------------
# 5. Testar novas mensagens
# -----------------------------

mensagens_teste = [
    "Clique agora para ganhar prêmio",
    "Vamos almoçar hoje",
    "Resgate seu dinheiro agora",
    "Te ligo mais tarde"
]

X_teste = vectorizer.transform(mensagens_teste)

predicoes = modelo.predict(X_teste)

print("CLASSIFICAÇÃO DAS NOVAS MENSAGENS:")
for msg, pred in zip(mensagens_teste, predicoes):
    print(f"{msg} -> {pred}")

print()

# -----------------------------
# 6. Palavras que mais indicam spam
# -----------------------------

spam_probs = modelo.feature_log_prob_[1]

df_spam = pd.DataFrame({
    "palavra": palavras,
    "peso_spam": spam_probs
})

print("PALAVRAS MAIS ASSOCIADAS A SPAM:")
print(df_spam.sort_values(by="peso_spam", ascending=False).head(10))
```
