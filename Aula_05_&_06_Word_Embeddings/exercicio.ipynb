```python
# Exercício - Treinamento de Word Embeddings com Word2Vec
# Objetivo: Treinar um modelo Word2Vec com sentenças de PLN e testar vocabulário

# Instalação das dependências (descomente se necessário)
# %pip install gensim matplotlib

# Importa o modelo Word2Vec da biblioteca Gensim
from gensim.models import Word2Vec
# Importa matplotlib para possível visualização dos vetores
import matplotlib.pyplot as plt

# Define o corpus de treino: lista de sentenças tokenizadas
# Cada sentença é uma lista de palavras (tokens)
sentencas = [
  ["o", "processamento", "de", "linguagem", "natural", "e", "incrivel"],
  ["aprendizado", "de", "maquina", "e", "uma", "subarea", "da", "ia"],
  ["redes", "neurais", "sao", "usadas", "em", "deep", "learning"],
  ["o", "gensim", "e", "excelente", "para", "word", "embeddings"]
]

# Treina o modelo Word2Vec com as sentenças
# vector_size=2: vetores de 2 dimensões (para facilitar visualização)
# window=2: contexto de 2 palavras ao redor
# min_count=1: mantém palavras que aparecem pelo menos 1 vez
model = Word2Vec(sentencas, vector_size=2, window=2, min_count=1)

# Testa se uma palavra fora do vocabulário é tratada corretamente
try:
    print(model.wv["gato"])
except KeyError:
    print("A palavra 'gato' não está no vocabulário do modelo.")
```
