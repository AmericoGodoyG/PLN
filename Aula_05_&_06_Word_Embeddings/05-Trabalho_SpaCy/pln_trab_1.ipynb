```python
# Trabalho 1 - Análise completa de texto com spaCy
# Objetivo: Tokenizar, lematizar, classificar (POS) e identificar stop words em uma notícia real

# Importa a biblioteca spaCy para processamento de linguagem natural
import spacy

# Carrega o modelo de português (versão pequena)
nlp = spacy.load("pt_core_news_sm")

# Texto de notícia real sobre a missão Artemis II da NASA
texto = """
O foguete Artemis II SLS (Space Launch System) da Nasa e a espaçonave Orion chegaram à plataforma de lançamento 39B, 
na manhã desta sexta-feira (20), após uma viagem de 11 horas desde o Edifício de Montagem de Veículos no Centro Espacial Kennedy da agência, na Flórida
O transportador sobre esteiras 2 da Nasa iniciou sua jornada de 6,4 quilômetros com o SLS e a espaçonave Orion integrados, acoplados à plataforma de lançamento móvel, às 00h20 (horário do leste dos EUA).
"""


# Processa o texto com o pipeline do spaCy
doc = nlp(texto)

# Itera sobre cada token exibindo: texto original, lema, classe gramatical (POS) e se é stop word
for token in doc:
    print(
        f"Texo: {token.text:10} | "
        f"Lemma: {token.lemma_:10}"
        f"POS: {token.pos_:10}"
        f"Stop: {token.is_stop}"
    )
```

```python
# Reconhecimento de Entidades Nomeadas (NER) do texto processado
# Exibe cada entidade encontrada e seu tipo (PER, ORG, LOC, MISC)
print("Entidades:")
for ent in doc.ents:
    print(ent.text, ent.label_)
```
