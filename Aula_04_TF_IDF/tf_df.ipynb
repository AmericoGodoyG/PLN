```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# 1. Definimos os documentos (o nosso corpus)
corpus = [
    "O gato comeu o peixe",
    "O gato dormiu no sofá",
    "O cachorro correu no parque"
]

# 2. Inicializamos o Vectorizer
# O parâmetro stop_words pode ser usado para igonarar
# palavras como "o", "no", etc.
vectorizer = TfidfVectorizer()

# 3. Ajustamos e transformamos os dados
tfidf_matrix = vectorizer.fit_transform(corpus)

# 4. Transformamos em um DateFrame para facilitar a vizualização
df_tfidf = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns = vectorizer.get_feature_names_out(),
    index = ['Documento 1', 'Documento 2', 'Documento 3']
)

# Define para não truncar as colunas (None significa sem limite)
pd.set_option('display.max_columns', 10000)
pd.set_option('display.precision', 2)

print(df_tfidf)
```
